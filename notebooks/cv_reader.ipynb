```python
# Import the PdfReader class from pypdf to read PDF files
from pypdf import PdfReader
```

```python
# Function to read and extract text from a PDF CV
def read_pdf(file_path):

    # Open the PDF document
    pdf = PdfReader(file_path)

    # Create an empty string to store the extracted text
    text = ""

    # Loop through all pages of the PDF document
    for page in pdf.pages:
        text += (page.extract_text() or "") + "\n"

    # Return the extracted text
    return text
```

```python
# Call our function using a sample CV
cv_text = read_pdf(ROOT / "samples" / "sample_cv.pdf")

# Display the extracted text
print(cv_text)
```

```python
# Import the Document class from python-docx to read Word (.docx) files
from docx import Document
```

```python
# Function to read and extract text from a Word CV
def read_word(file_path):

    # Open the Word document
    word = Document(file_path)

    # Create an empty string to store the extracted text
    text = ""

    # Loop through all paragraphs of the Word document
    for paragraph in word.paragraphs:
        text += paragraph.text + "\n"

    # Return the extracted text
    return text
```

```python
# Test the Word CV reader
cv_text = read_word(ROOT / "samples" / "sample_cv.docx")

# Display the extracted text
print(cv_text)
```

```python
import pymupdf

# Function to extract text from a PDF CV
def read_pdf(file_path):

    # Open the PDF document
    pdf = pymupdf.open(file_path)

    # Store the extracted text
    text = ""

    # Read every page of the PDF
    for page in pdf:
        text += page.get_text() + " "

    # Close the PDF
    pdf.close()

    # Return the extracted text
    return text
    
```

```python
# Test the PDF reader function
cv_text = read_pdf(ROOT / "samples" / "sample_cv.pdf")

# Display the extracted text
print(cv_text)
```

```python
# Import the library for reading Word documents
from docx import Document

# Function to extract text from a Word CV
def read_docx(file_path):

    # Open the Word document
    doc = Document(file_path)

    # Store the extracted text
    text = ""

    # Read every paragraph in the document
    for paragraph in doc.paragraphs:
        text += paragraph.text + " "

    # Return the extracted text
    return text
```

```python
# Function to read and extract text from a plain-text CV
def read_txt(file_path):

    # UTF-8 supports letters from different languages
    with open(file_path, "r", encoding="utf-8") as file:
        text = file.read()

    # Return the extracted text
    return text
```

```python
# Test the TXT CV reader function
cv_text = read_txt(ROOT / "samples" / "sample_cv.txt")

# Display the extracted text from the TXT file
print(cv_text)
```
