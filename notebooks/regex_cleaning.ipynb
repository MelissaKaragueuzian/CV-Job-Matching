```python
# Find the project folder whether this notebook runs from the
# project root or from the notebooks folder.
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "samples").exists():
    ROOT = ROOT.parent
```

```python
# Import the regular expressions library
import re
```

```python
# Create a function that receives CV text
def remove_emails(text):

    # Find emails and replace them with a space
    cleaned_text = re.sub(
        # \b: word boundary
        # [\w.+-]+: username (one or more characters)
        # @: separates username and domain
        # [\w.-]+: domain name
        # \.: matches a dot
        # [A-Za-z]{2,}: extension with at least 2 letters
        # \b: ending word boundary
        r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b",
        " ",
        text
    )

    # Return the text without email addresses
    return cleaned_text
```

```python
# Sample CV text containing an email address
sample = "My name is Melissa. Email: melissa@gmail.com. Skills: Python, SQL."

# Call the function to remove the email
result = remove_emails(sample)

# Display the cleaned text
print(result)
```

```python
# Function to remove website links from CV text
def remove_urls(text):
    # https? : matches http or https
    # ://    : matches the characters ://
    # \S+   : matches one or more non-whitespace characters
    cleaned_text = re.sub(
        r"https?://\S+",
        " ",
        text
    )

    # Return the text without website links
    return cleaned_text
```

```python
# Sample CV containing a website link
sample = "I know Python. Portfolio: https://example.com/projects Skills: SQL."

# Call the function
result = remove_urls(sample)

# Display the result
print(result)
```

```python
# Function to remove extra whitespace from CV text
def remove_extra_spaces(text):

    # \s: matches whitespace (spaces, tabs, newlines)
    # +: matches one or more repetitions
    cleaned_text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Remove spaces at the beginning and end
    return cleaned_text.strip()
```

```python
# Sample text containing unnecessary spaces
sample = "Python     SQL      Machine Learning"

# Call the function
result = remove_extra_spaces(sample)

# Display the result
print(result)
```

```python
# Function to clean CV text using all three regex functions
def clean_text(text):

    # Step 1: Remove email addresses
    text = remove_emails(text)

    # Step 2: Remove website URLs
    text = remove_urls(text)

    # Step 3: Remove extra whitespace
    text = remove_extra_spaces(text)

    # Return the final cleaned text
    return text
```

```python
# Sample CV containing emails, URLs and extra spaces
sample = """
My name is Melissa.
Email: melissa@gmail.com
Portfolio: https://example.com

Skills: Python,   SQL,    Machine Learning
Experience: 2 years
"""

# Clean the entire CV text
result = clean_text(sample)

# Display the final result
print(result)
```

```python
# Read the CV from a TXT file
with open(ROOT / "samples" / "sample_cv.txt", "r", encoding="utf-8") as file:
    cv_text = file.read()

# Apply our regex cleaning function
cleaned_cv = clean_text(cv_text)

# Display the original CV
print("BEFORE CLEANING:")
print(cv_text)

# Display the cleaned CV
print("\nAFTER CLEANING:")
print(cleaned_cv)
```

```python
# Import the library to read PDF files
import pymupdf

# Open the PDF CV
pdf = pymupdf.open(ROOT / "samples" / "sample_cv.pdf")

# Extract text from every page
cv_text = ""

for page in pdf:
    cv_text += page.get_text() + " "

# Close the PDF file
pdf.close()

# Clean the extracted text using our regex function
cleaned_cv = clean_text(cv_text)

# Display the result
print("BEFORE CLEANING:")
print(cv_text)

print("\nAFTER CLEANING:")
print(cleaned_cv)
```

```python
# Import the library to read Word documents
from docx import Document

# Open the Word CV
doc = Document(ROOT / "samples" / "sample_cv.docx")

# Extract text from all paragraphs
cv_text = ""

for paragraph in doc.paragraphs:
    cv_text += paragraph.text + " "

# Apply our regex cleaning function
cleaned_cv = clean_text(cv_text)

# Display the original and cleaned text
print("BEFORE CLEANING:")
print(cv_text)

print("\nAFTER CLEANING:")
print(cleaned_cv)
```
