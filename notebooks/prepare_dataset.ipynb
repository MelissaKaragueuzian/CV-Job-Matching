```python
# Find the project folder whether this notebook runs from the
# project root or from the notebooks folder.
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "samples").exists():
    ROOT = ROOT.parent
```

```python
# Import the Pandas library to work with datasets
# 'pd' is an alias (short name) for pandas
import pandas as pd
```

```python
# Store the paths of the two dataset versions:
# original : real job listings
# augmented : additional generated job listings
splits = {'original': 'data/original-00000-of-00001.parquet', 
          'augmented': 'data/augmented-00000-of-00001.parquet'}

# Load the original job dataset from Hugging Face
# read_parquet() is a function that reads the dataset into a Pandas DataFrame
df = pd.read_parquet("hf://datasets/its-zion-18/Jobs-tabular-dataset/" 
                    + splits["original"])
```

```python
# Display the first 5 rows of the dataset
# This help us how the data is organized
df.head()
```

```python
# Display the dataset dimensions (rows, columns)
# Our dataset contains 391 jobs and 7 columns
df.shape
```

```python
# Check for missing (null) values in each column
# isnull() identifies missing values
# sum() counts them for each column
df.isnull().sum()
```

```python
# Convert the number of duplicates rows
duplicates = df.duplicated().sum()

# Convert the result to a normal Python integer
print(int(duplicates))
```

```python
# Remove duplicate rows from the dataset
df = df.drop_duplicates()

# Check the new dataset size
print(df.shape)
```

```python
# Combine job information into one organized text column
# Keep the original dataset column names as labels

df["job_text"] = (
    "Job title: " + df["Job title"] + "\n" +
    "Description: " + df["Description"] + "\n" +
    "qualifications: " + df["qualifications"] + "\n" +
    "Duties: " + df["Duties"]
)

# iloc[0] selects the first row by position
print(df["job_text"].iloc[0])
```

```python
# Check the updated dataset dimensions
print(df.shape)

# Save the prepared dataset as a JSON file
# orient='records' stores one JSON object per job
df.to_json(
    ROOT / "data" / "jobs_clean.json",
    orient="records",
    force_ascii=False,
    indent=2,
)

print("Dataset saved successfully!")
```
